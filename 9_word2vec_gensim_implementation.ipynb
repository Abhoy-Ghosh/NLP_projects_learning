

# Word2Vec (According to Your Notes)

### Definition

> Word2Vec is a neural network-based word embedding technique developed by Google (2013) that converts words into dense vectors while capturing semantic relationships between words. 

>**Word2Vec** is a neural network-based technique that converts words into dense numerical vectors (embeddings) while preserving semantic and contextual relationships between words.

---

## Why Word2Vec?

Previous techniques:

```text
One Hot Encoding
↓
Bag of Words
↓
TF-IDF
```

Problems:

* Sparse matrix
* No semantic meaning
* Large vocabulary size
* OOV (Out Of Vocabulary) issues

Word2Vec solves these by learning **dense vector representations**. 

---

## Word Embedding

Word → Dense Vector

```text
King  → [-0.92, 0.95, 0.97, ...]
Queen → [ 0.93, 0.96, 0.68, ...]
```

Words with similar meanings are located closer in vector space. 

---

## Types of Word2Vec

### 1. CBOW (Continuous Bag of Words)

Predict target word from surrounding context.

```text
Context Words
      ↓
     CBOW
      ↓
Target Word
```

Example:

```text
Infosys Company Related To Data Science

Input:
Infosys, Company, Related, To

Output:
IS
```

CBOW works better on smaller datasets. 

---

### 2. Skip-Gram

Predict surrounding words from target word.

```text
Target Word
      ↓
   Skip-Gram
      ↓
Context Words
```

Example:

```text
Input:
IS

Output:
Infosys, Company, Related, To
```

Skip-Gram performs better on large datasets. 

---

## Window Size

Defines how many neighboring words are considered.

Example:

```text
Infosys Company Related To Data Science

Window Size = 2
```

For word:

```text
Related
```

Context:

```text
Company, Related, To
```

Increasing window size generally improves contextual understanding but increases training complexity. 

---

## Feature Representation

Word2Vec learns features automatically:

```text
King
↓
Gender
Royal
Age
Food
...
```

Instead of manually creating features. 

---

## Semantic Relationships

Word2Vec can perform vector arithmetic:

```text
King - Man + Woman = Queen
```

or

```text
King - Boy + Girl = Queen
```

This shows semantic information is captured in embeddings. 

---

## Cosine Similarity

Measures similarity between two vectors.

Formula:

\cos(\theta)=\frac{A\cdot B}{|A||B|}

Interpretation:

```text
1   → Exactly Similar
0   → Unrelated
-1  → Opposite
```

Used extensively with Word2Vec. 

---

## Advantages

* Dense matrix instead of sparse matrix
* Captures semantic meaning
* Fixed dimensional representation (e.g., 300 dimensions)
* Similar words are placed close together
* Better than BoW and TF-IDF for semantic tasks

# Average Word2Vec

### Problem

Word2Vec gives vectors for individual words.

Example:

```text
The food is good
```

Need one vector for the entire sentence/document.

---

## Solution: Average Word2Vec

Take the average of all word vectors.

```text
Word Vectors
      ↓
Average
      ↓
Sentence Vector
```

Example:

```text
The food is good
```

```text
The   → Vector1
food  → Vector2
is    → Vector3
good  → Vector4
```

Average:

```text
(Vector1 + Vector2 + Vector3 + Vector4) / 4
```

Result:

```text
300-Dimensional Document Vector
```

This single vector can be fed into ML algorithms. 

---

## Workflow

```text
Text
 ↓
Pretrained Word2Vec (Google/Gensim)
 ↓
Word Embeddings
 ↓
Average All Vectors
 ↓
Document Vector
 ↓
Machine Learning Model
```

---

## Why Use Average Word2Vec?

* Fixed-size representation
* Better semantic representation than BoW
* Easy to implement
* Works well for classification tasks

````python
# avgword2vec 

import numpy as np

def avg_word2vec(words, model):

    vectors = [
        model[word]
        for word in words
        if word in model
    ]

    if len(vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(vectors, axis=0)

avg_word2vec("I love nlp".split(),wv_model)
````
