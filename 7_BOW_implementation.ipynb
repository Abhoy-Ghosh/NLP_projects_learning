## Bags Of Words 

````python
import pandas as pd
import numpy as np
````

````python
messages = pd.read_csv("data/SMSSpamCollection.csv",
                       sep = "\t", names = ["label","message"])

messages
````

````python
## Data Cleaning and prepropressing

import re # impprting regular expression
import nltk 

````

````python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

ps=PorterStemmer()
````

```
Raw Text
    ↓
Regex Cleaning
    ↓
Lowercase
    ↓
Tokenization (split)
    ↓
Stopword Removal
    ↓
Stemming
    ↓
Join Back
    ↓
Corpus 
```

````python
corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if  word not in  stopwords.words()]
    review = ' '.join(review)
    corpus.append(review)



````
