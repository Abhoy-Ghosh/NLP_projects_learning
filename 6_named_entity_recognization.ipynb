# Named Entity Recognition (NER)

### Definition

> **Named Entity Recognition (NER)** is an NLP technique used to identify and classify named entities in text into predefined categories such as Person, Location, Organization, Date, Time, Money, etc.

---

### Example

**Sentence:**

```text
Elon Musk is the CEO of Tesla and lives in Texas.
```

**NER Output:**

| Entity    | Type         |
| --------- | ------------ |
| Elon Musk | PERSON       |
| Tesla     | ORGANIZATION |
| Texas     | LOCATION     |

---

### Common Entity Types

| Entity Type    | Example                |
| -------------- | ---------------------- |
| PERSON         | Virat Kohli, Elon Musk |
| ORGANIZATION   | Google, Microsoft      |
| LOCATION (GPE) | India, Kolkata         |
| DATE           | 15 August 2025         |
| TIME           | 10:30 AM               |
| MONEY          | $1000                  |
| PERCENT        | 25%                    |
| PRODUCT        | iPhone 16              |
| EVENT          | FIFA World Cup         |

---

### NLP Pipeline

```text
Text
 ↓
Tokenization
 ↓
POS Tagging
 ↓
Named Entity Recognition
 ↓
Extract Entities
```

---

### NLTK Example

```python
import nltk

sentence = "Elon Musk founded SpaceX in California"

words = nltk.word_tokenize(sentence)

pos_tags = nltk.pos_tag(words)

nltk.ne_chunk(pos_tags)
```

Output (Tree):

```text
(PERSON Elon Musk)
(ORGANIZATION SpaceX)
(GPE California)
```

### Difference Between POS Tagging and NER

| POS Tagging                 | NER                            |
| --------------------------- | ------------------------------ |
| Identifies grammatical role | Identifies real-world entities |
| Elon → NNP                  | Elon Musk → PERSON             |
| Google → NNP                | Google → ORGANIZATION          |
