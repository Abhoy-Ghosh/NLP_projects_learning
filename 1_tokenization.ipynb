````python
corpus = """Hello, welcome to nlp journey. It is the tokenzation example code!
Tokenization is the process of breaking a text into smaller units called tokens (words, sentences, or subwords), so that it can be processed by NLP models."""
````

````python
print(corpus)
````

````python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
````


## sent_tokenize() uses the Punkt Sentence Tokenizer internally.
```
sent_tokenize()
       ↓
PunktTokenizer
       ↓
punkt_tab resource
For your NLP notes:
```
### punkt: 
Pre-trained sentence boundary detection model used by NLTK for sentence tokenization.

### punkt_tab:
 Supporting tokenizer resource required by newer NLTK versions for sentence splitting.

````python
# Tokenization
#  paragraphs - >  sentence

from nltk.tokenize import sent_tokenize

documents = sent_tokenize(corpus)
print(documents)
````

````python
for sentence in documents:
    print("-",sentence,"-")
````

````python
# paragraphs -> word

from nltk.tokenize import word_tokenize

word_tokenize(corpus)
````

````python
# sentencce -> word 

for sentence in documents:
    print("-",word_tokenize(sentence),"-")
````

````python
ex1 = "I am NLP"
ex2 = "I'm NLP"
````

````python
from nltk.tokenize import wordpunct_tokenize
print(word_tokenize(ex2))
print(wordpunct_tokenize(ex2))
````

````python
from nltk.tokenize import  TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
# only last full stop is counted in separarted as token among in all full stops

tokenizer.tokenize(corpus)
````
