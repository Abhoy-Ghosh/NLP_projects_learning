**Morphology** in NLP and linguistics is the study of the **structure of words and how words are formed from smaller meaningful units called morphemes**.

### Simple Definition

> **Morphology is the branch of linguistics that studies the formation, structure, and components of words.**

---

### Example

Consider the word:

```text
playing
```

It can be broken into:

```text
play + ing
```

* **play** → root word
* **ing** → suffix

Here:

```text
playing → play
```

The process of identifying the root form is called **morphological analysis**.

---

### More Examples

| Word    | Morphological Parts | Root/Lemma |
| ------- | ------------------- | ---------- |
| playing | play + ing          | play       |
| studies | study + es          | study      |
| better  | irregular form      | good       |
| running | run + ing           | run        |

---

### How WordNet Lemmatizer Uses Morphology

The WordNet Lemmatizer uses a function called **morphy()**.

Example:

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("running", pos='v'))
```

Output:

```text
run
```

What happens internally?

```text
running
    ↓
morphological analysis
    ↓
remove suffix "ing"
    ↓
find dictionary form
    ↓
run
```

---

### Stemming vs Morphological Lemmatization

| Stemming                  | Lemmatization                |
| ------------------------- | ---------------------------- |
| Uses simple rules         | Uses morphology + dictionary |
| Faster                    | Slower                       |
| May produce invalid words | Produces valid words         |
| studies → studi           | studies → study              |
| caring → care             | caring → care                |

---

### Note for Your NLP Notes

> **Morphology:** The study of word formation and structure. It analyzes how words are built from roots, prefixes, and suffixes to determine their base or dictionary form (lemma).

> **Morphological Analysis:** The process of identifying the root word and grammatical components (prefixes, suffixes, inflections) of a word.

Example:

```text
running → run
studies → study
better → good
```

This is why **WordNetLemmatizer is called a morphological lemmatizer**—it uses morphological rules and a dictionary (WordNet) to find the correct lemma.



## Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −

````python
import nltk
nltk.download('wordnet')
````

````python

## Q&A,chatbots,text summarization
from nltk.stem import WordNetLemmatizer
````

````python
lemmatizer = WordNetLemmatizer()
````

````python

'''
POS- 
Noun-n (by default)
verb-v
adjective-a
adverb-r
'''
print(lemmatizer.lemmatize("going"))
print(lemmatizer.lemmatize("going",pos='n'))
print(lemmatizer.lemmatize("going",pos='v')) # only this work correctly as it is a verb 
print(lemmatizer.lemmatize("going",pos='a'))
print(lemmatizer.lemmatize("going",pos='r'))

````

````python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
````
